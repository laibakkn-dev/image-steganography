```python
from PIL import Image

# ── STEP 1: Convert text to binary ──────────────────────────────────
def text_to_binary(text):
    binary = ""
    for char in text:
        binary += format(ord(char), '08b')
    return binary

# ── STEP 2: Hide message inside image ───────────────────────────────
def encode(image_path, message, output_path):
    img = Image.open(image_path)
    img = img.convert("RGB")
    pixels = list(img.getdata())

    message = message + "###"
    binary_message = text_to_binary(message)

    index = 0
    new_pixels = []

    for pixel in pixels:
        r, g, b = pixel
        new_pixel = list(pixel)

        for i in range(3):
            if index < len(binary_message):
                new_pixel[i] = (new_pixel[i] & 0b11111110) | int(binary_message[index])
                index += 1

        new_pixels.append(tuple(new_pixel))

    new_img = Image.new("RGB", img.size)
    new_img.putdata(new_pixels)
    new_img.save(output_path)
    print("Message hidden successfully! Saved as:", output_path)

# ── STEP 3: Extract message from image ──────────────────────────────
def decode(image_path):
    img = Image.open(image_path)
    img = img.convert("RGB")
    pixels = list(img.getdata())

    binary_message = ""
    for pixel in pixels:
        for channel in pixel:
            binary_message += str(channel & 1)

    message = ""
    for i in range(0, len(binary_message), 8):
        byte = binary_message[i:i+8]
        char = chr(int(byte, 2))
        message += char
        if message.endswith("###"):
            break

    return message[:-3]

# ── MAIN MENU ────────────────────────────────────────────────────────
def main():
    print("================================")
    print("  IMAGE STEGANOGRAPHY TOOL")
    print("  Author     : [Laiba jalil]")
    print("  Supervisor : [Mr Ghulam Ali]")
    print("================================")

    while True:                           # keeps looping until user exits
        print("\n1. Hide a message in image")
        print("2. Read hidden message from image")
        print("3. Exit")
        choice = input("\nEnter choice (1, 2 or 3): ")

        if choice == "1":
            image  = input("Enter image filename (e.g. photo.png): ")
            msg    = input("Enter secret message: ")
            output = input("Enter output filename (e.g. output.png): ")
            encode(image, msg, output)

        elif choice == "2":
            image = input("Enter stego image filename: ")
            result = decode(image)
            print("\nHidden message is:", result)

        elif choice == "3":
            print("Goodbye!")
            break                         # exits the loop

        else:
            print("Invalid choice! Please enter 1, 2 or 3.")

main()
```
